```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 1000  # number of records

df = pd.DataFrame({
    'Order_ID': range(1, n+1),
    'Order_Date': pd.date_range(start='2023-01-01', periods=n, freq='D'),
    'Region': np.random.choice(['North', 'South', 'East', 'West'], n),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Furniture'], n),
    'Customer_Segment': np.random.choice(['Consumer', 'Corporate', 'Home Office'], n),
    'Sales': np.random.randint(100, 20000, n),
    'Discount': np.round(np.random.uniform(0, 0.4, n), 2),
    'Profit': np.random.randint(-2000, 5000, n),
    'Delivery_Days': np.random.randint(1, 12, n),
    'Payment_Mode': np.random.choice(['Cash', 'Card', 'UPI'], n),
    'Customer_Rating': np.random.randint(1, 6, n)
})

# Save dataset
df.to_csv('urbankart_transactions.csv', index=False)

df.head()
```

```python
df = pd.read_csv('urbankart_transactions.csv')
df.head()
```

```python
df.info()
df.describe()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

```python
df.isnull().sum()
```

```python
# Example
df['Sales'].fillna(df['Sales'].median(), inplace=True)
df['Profit'].fillna(df['Profit'].median(), inplace=True)
df['Customer_Rating'].fillna(df['Customer_Rating'].mode()[0], inplace=True)
```

```python
df['Region'] = df['Region'].str.lower().str.strip()
df['Payment_Mode'] = df['Payment_Mode'].str.lower().str.strip()
```

```python
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df['Sales'] < Q1 - 1.5*IQR) | (df['Sales'] > Q3 + 1.5*IQR)]
len(outliers)
```

```python
df['Sales'].skew()
df['Profit'].skew()
```

```python
df['Sales'].mean(), df['Sales'].median()
df['Profit'].mean(), df['Profit'].median()
```

```python
std_sales = df['Sales'].std()
mean_sales = df['Sales'].mean()
cv = std_sales / mean_sales

std_sales, cv
```

```python
import scipy.stats as stats
import numpy as np

mean = df['Profit'].mean()
std = df['Profit'].std()
n = len(df)

ci = stats.t.interval(0.95, n-1, loc=mean, scale=std/np.sqrt(n))
ci
```

```python
grouped = df.groupby(['Region','Category','Customer_Segment']).agg({
    'Sales':['mean','median'],
    'Profit':'mean',
    'Order_ID':'count'
}).reset_index()

grouped
```

```python
global_sales = df['Sales'].mean()
global_profit = df['Profit'].mean()
```

```python
grouped[grouped[('Sales','mean')] > global_sales]
grouped[grouped[('Profit','mean')] < global_profit]
```
